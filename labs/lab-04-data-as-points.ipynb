## 2. Many Objects as a Data Matrix

A dataset with many objects becomes a matrix.

Rows usually represent objects. Columns usually represent features.

```python
# columns: hours studied, sleep hours, exam score
X = np.array([
    [2, 7, 70],
    [5, 6, 85],
    [1, 8, 65],
    [7, 5, 92],
    [4, 7, 80],
    [6, 6, 88],
    [3, 8, 75]
])

X, X.shape
```

```python
# Row view: one object
X[1]
```

```python
# Column view: one feature across all objects
exam_scores = X[:, 2]
exam_scores
```

## 3. Visualizing Two Features

To visualize a high-dimensional row, we can choose two features and make a scatter plot.

Here we plot hours studied versus exam score.

```python
hours = X[:, 0]
scores = X[:, 2]

plt.scatter(hours, scores, s=90)
for i, (h, s) in enumerate(zip(hours, scores), start=1):
    plt.text(h + 0.06, s, f"S{i}")
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("Students as Points in Feature Space")
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. The Mean Vector and Centering

The mean vector is the coordinate-wise average of the data points.

Centering subtracts the mean vector from every row.

```python
mean_vector = X.mean(axis=0)
X_centered = X - mean_vector

print("Mean vector:", mean_vector)
print("Mean of centered data:", X_centered.mean(axis=0))
X_centered
```

```python
plt.scatter(X[:, 0], X[:, 2], s=90, label="original")
plt.scatter(mean_vector[0], mean_vector[2], s=200, marker="x", label="mean")
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("Mean Vector as the Center of the Cloud")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.scatter(X_centered[:, 0], X_centered[:, 2], s=90)
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.xlabel("Centered hours")
plt.ylabel("Centered score")
plt.title("Centered Cloud")
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. A Larger Data Cloud with a Trend

Synthetic data is useful because we know the story that generated it.

Here we create a data cloud with an upward trend.

```python
np.random.seed(1)

hours = np.random.uniform(0, 8, 100)
scores = 55 + 5 * hours + np.random.normal(0, 6, size=100)

plt.scatter(hours, scores, alpha=0.75)
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("A Data Cloud with a Trend")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Fit a line only to summarize the trend.
m, b = np.polyfit(hours, scores, 1)
print("slope:", m)
print("intercept:", b)

x_line = np.linspace(0, 8, 100)
y_line = m * x_line + b

plt.scatter(hours, scores, alpha=0.6)
plt.plot(x_line, y_line, linewidth=2)
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("Trend Line Through a Data Cloud")
plt.grid(True, alpha=0.3)
plt.show()
```

## 6. Scaling Changes Geometry

Distance depends on units.

A feature measured in large numbers can dominate the geometry.

```python
houses = np.array([
    [1400, 430000],
    [1800, 500000],
    [1900, 510000],
    [2500, 700000],
    [1200, 350000]
])

raw_distance = np.linalg.norm(houses[1] - houses[2])
print("Raw distance between house 2 and house 3:", raw_distance)
```

```python
means = houses.mean(axis=0)
stds = houses.std(axis=0)
houses_scaled = (houses - means) / stds

scaled_distance = np.linalg.norm(houses_scaled[1] - houses_scaled[2])
print("Scaled distance between house 2 and house 3:", scaled_distance)
houses_scaled
```

```python
fig, ax = plt.subplots()
ax.scatter(houses[:, 0], houses[:, 1], s=100)
for i, p in enumerate(houses, start=1):
    ax.text(p[0] + 20, p[1], f"H{i}")
ax.set_xlabel("Size in square feet")
ax.set_ylabel("Price in dollars")
ax.set_title("Raw House Feature Space")
ax.grid(True, alpha=0.3)
plt.show()

fig, ax = plt.subplots()
ax.scatter(houses_scaled[:, 0], houses_scaled[:, 1], s=100)
for i, p in enumerate(houses_scaled, start=1):
    ax.text(p[0] + 0.03, p[1], f"H{i}")
ax.set_xlabel("Standardized size")
ax.set_ylabel("Standardized price")
ax.set_title("Scaled House Feature Space")
ax.grid(True, alpha=0.3)
plt.show()
```

## 7. Clusters

Clusters are groups of nearby points.

We create three customer groups: coffee lovers, tea lovers, and customers who buy both.

```python
np.random.seed(5)
coffee = np.random.normal(loc=[8, 2], scale=[0.8, 0.8], size=(50, 2))
tea = np.random.normal(loc=[2, 8], scale=[0.8, 0.8], size=(50, 2))
both = np.random.normal(loc=[6, 6], scale=[0.8, 0.8], size=(50, 2))
customers = np.vstack([coffee, tea, both])
labels = np.array(["coffee"] * 50 + ["tea"] * 50 + ["both"] * 50)

plt.scatter(customers[:, 0], customers[:, 1], alpha=0.75)
plt.xlabel("Coffee purchases")
plt.ylabel("Tea purchases")
plt.title("Customer Data Cloud")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
centers = np.array([coffee.mean(axis=0), tea.mean(axis=0), both.mean(axis=0)])
centers
```

```python
plt.scatter(customers[:, 0], customers[:, 1], alpha=0.5)
plt.scatter(centers[:, 0], centers[:, 1], s=250, marker="x")
for name, c in zip(["coffee", "tea", "both"], centers):
    plt.text(c[0] + 0.1, c[1], name)
plt.xlabel("Coffee purchases")
plt.ylabel("Tea purchases")
plt.title("Cluster Centers")
plt.grid(True, alpha=0.3)
plt.show()
```

## 8. Outliers

Outliers are points far from the main pattern.

They can be errors, unusual cases, or important discoveries.

```python
np.random.seed(8)
normal_points = np.random.normal(loc=[0, 0], scale=1, size=(100, 2))
outliers = np.array([[5, 5], [-5, 4], [4, -5]])
X_out = np.vstack([normal_points, outliers])

center = X_out.mean(axis=0)
dist_to_center = np.linalg.norm(X_out - center, axis=1)
threshold = np.percentile(dist_to_center, 97)
flagged = dist_to_center >= threshold

plt.scatter(X_out[:, 0], X_out[:, 1], alpha=0.6, label="points")
plt.scatter(X_out[flagged, 0], X_out[flagged, 1], s=180, marker="x", label="flagged")
plt.scatter(center[0], center[1], s=150, marker="*", label="mean")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Simple Outlier Flagging by Distance to Center")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 9. Nearest Neighbor Classification

A new point can be classified by looking at nearby labeled points.

```python
np.random.seed(11)
A = np.random.normal(loc=[2.0, 0.7], scale=[0.25, 0.12], size=(40, 2))
B = np.random.normal(loc=[5.0, 1.6], scale=[0.35, 0.18], size=(40, 2))
X_train = np.vstack([A, B])
y_train = np.array([0] * len(A) + [1] * len(B))
new_point = np.array([3.7, 1.15])

distances = np.linalg.norm(X_train - new_point, axis=1)
nearest = np.argmin(distances)
print("Nearest label:", "B" if y_train[nearest] == 1 else "A")
print("Nearest distance:", distances[nearest])

plt.scatter(A[:, 0], A[:, 1], label="Class A")
plt.scatter(B[:, 0], B[:, 1], label="Class B")
plt.scatter(new_point[0], new_point[1], s=200, marker="*", label="new point")
plt.plot([new_point[0], X_train[nearest,0]], [new_point[1], X_train[nearest,1]], linewidth=2)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Nearest Neighbor Classification")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 10. Images as Points

An image can be flattened into a vector.

Below, each $8 \times 8$ image becomes a point in $\mathbb{R}^{64}$.

```python
def make_digit_like(seed, shift=0):
    rng = np.random.default_rng(seed)
    img = np.zeros((8, 8))
    img[1:7, 2 + shift] = 1
    img[1:7, 5 + shift] = 1
    img[1, 2 + shift:6 + shift] = 1
    img[6, 2 + shift:6 + shift] = 1
    img += rng.normal(0, 0.15, size=(8, 8))
    return np.clip(img, 0, 1)

images = [make_digit_like(i, shift=0) for i in range(4)]

fig, axes = plt.subplots(1, 4, figsize=(10, 3))
for ax, img in zip(axes, images):
    ax.imshow(img, cmap="gray_r", vmin=0, vmax=1)
    ax.axis("off")
plt.suptitle("Small Images")
plt.show()

vectors = np.array([img.reshape(-1) for img in images])
vectors.shape
```

```python
# Pairwise distances between image vectors
D = np.zeros((len(vectors), len(vectors)))
for i in range(len(vectors)):
    for j in range(len(vectors)):
        D[i, j] = np.linalg.norm(vectors[i] - vectors[j])
D
```

## 11. High-Dimensional Geometry

We now explore random points in different dimensions.

Watch how average distance changes as dimension increases.

```python
np.random.seed(21)
dims = [2, 5, 10, 20, 50, 100, 200, 500]
means = []
stds = []
relative_stds = []

for d in dims:
    X1 = np.random.normal(size=(1000, d))
    X2 = np.random.normal(size=(1000, d))
    dist = np.linalg.norm(X1 - X2, axis=1)
    means.append(dist.mean())
    stds.append(dist.std())
    relative_stds.append(dist.std() / dist.mean())

plt.plot(dims, means, marker="o")
plt.xlabel("Dimension")
plt.ylabel("Mean distance")
plt.title("Average Distance Increases with Dimension")
plt.grid(True, alpha=0.3)
plt.show()

plt.plot(dims, relative_stds, marker="o")
plt.xlabel("Dimension")
plt.ylabel("Standard deviation / mean")
plt.title("Distances Become More Concentrated")
plt.grid(True, alpha=0.3)
plt.show()
```
